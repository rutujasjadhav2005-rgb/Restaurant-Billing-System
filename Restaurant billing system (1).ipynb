```python
#RESTAURANT BILLING SYSTEM

# Menu Item Class
class MenuItem:

    def __init__(self, name, price):
        self.name = name
        self.price = price

    def display_item(self):
        print(self.name, " - ₹", self.price)

# Class
class Restaurant:
    #Cunstructor
    def __init__(self):

        # Dictionary
        self.menu = {   #Parameter
            "Pizza": 250,
            "Burger": 150,
            "Pasta": 180,
            "Sandwich": 100,
            "Coffee": 80,
            "Cold Drink": 60
        }
        # List to store orders
        self.orders = []

#menu
    def display_menu(self):

        print("\n========== RESTAURANT MENU ==========")

        for item in self.menu:

            print(item, " - ₹", self.menu[item])

#add order
    def add_order(self):

        self.display_menu()

        item = input("\nEnter item name: ")

        if item in self.menu:

            quantity = int(input("Enter quantity: "))

            price = self.menu[item]

            total = price * quantity

            order = {
                "item": item,
                "price": price,
                "quantity": quantity,
                "total": total
            }

            self.orders.append(order)

            print("\nOrder added successfully!")

        else:

            print("\nItem not available!")

#display order
    def display_order(self):

        if len(self.orders) == 0:

            print("\nNo orders available.")

        else:

            print("\n========== YOUR ORDER ==========")

            print("Item\tPrice\tQty\tTotal")
            print("--------------------------------")

            for order in self.orders:

                print(
                    order["item"],
                    "\t",
                    order["price"],
                    "\t",
                    order["quantity"],
                    "\t",
                    order["total"]
                )

#calculate subtotal
    def calculate_subtotal(self):

        subtotal = 0

        for order in self.orders:

            subtotal = subtotal + order["total"]

        return subtotal

#calculate GST
    def calculate_gst(self):

        subtotal = self.calculate_subtotal()

        gst = subtotal * 0.05

        return gst

#calculate final bill
    def calculate_final_bill(self):

        subtotal = self.calculate_subtotal()

        gst = self.calculate_gst()

        final_bill = subtotal + gst

        return final_bill

#display bill

    def display_bill(self):

        if len(self.orders) == 0:

            print("\nNo orders available.")

        else:

            self.display_order()

            subtotal = self.calculate_subtotal()

            gst = self.calculate_gst()

            final_bill = self.calculate_final_bill()

            print("\n========== BILL ==========")

            print("Subtotal    : ₹", subtotal)
            print("GST (5%)    : ₹", gst)
            print("Final Bill  : ₹", final_bill)

#Object
restaurant = Restaurant()

# Main Menu
while True:

    print("\n====================================")
    print("       RESTAURANT BILLING SYSTEM")
    print("====================================")

    print("1. Display Menu")
    print("2. Add Order")
    print("3. Display Order")
    print("4. Calculate Subtotal")
    print("5. Calculate GST")
    print("6. Calculate Final Bill")
    print("7. Exit")


    choice = int(input("\nEnter your choice: "))

    if choice == 1:
        restaurant.display_menu()

    elif choice == 2:
        restaurant.add_order()

    elif choice == 3:
        restaurant.display_order()
        
    elif choice == 4:
        subtotal = restaurant.calculate_subtotal()
        print("\nSubtotal = ₹", subtotal)

    elif choice == 5:
        gst = restaurant.calculate_gst()
        print("\nGST = ₹", gst)
        
    elif choice == 6:
        final_bill = restaurant.calculate_final_bill()
        print("\nFinal Bill = ₹", final_bill)

    elif choice == 7:
        print("\nThank you for visiting!")
        break

    else:
        print("\nInvalid choice! Please enter 1 to 7.")
```
